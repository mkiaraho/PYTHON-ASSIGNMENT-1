```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('sales_data.csv')
```

```python
revenue_trend = df.groupby('Date')['Revenue'].sum()

```

```python
plt.figure(figsize=(12,6))
plt.plot(revenue_trend, color='blue')
plt.title('Daily Revenue Trend')
plt.xlabel('Date')
plt.ylabel('Total Revenue (Ksh)')
plt.grid(True)
plt.show()
```

```python
avg_cost_per_category = df.groupby('Product_Category')['Unit_Cost'].mean()

```

```python
avg_cost_per_category.plot(kind='bar', color='orange', figsize=(8,5))
plt.title('Average Unit Cost by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Average Unit Cost (Ksh)')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Order_Quantity'], bins=15, color='green', edgecolor='black')
plt.title('Distribution of Order Quantity')
plt.xlabel('Order Quantity')
plt.ylabel('Frequency')
plt.grid(axis='y')
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(df['Profit'], df['Revenue'], color='purple', alpha=1)
plt.title('Relationship between Profit and Revenue')
plt.xlabel('Profit (Ksh)')
plt.ylabel('Revenue (Ksh)')
plt.grid(True)
plt.show()
```
